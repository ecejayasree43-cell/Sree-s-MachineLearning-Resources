```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.preprocessing import LabelEncoder

# Data
emails=[
    "win money",
    "free prize",
    "hello friend",
    "how are you"

]

labels=["spam","spam","Not spam","Not spam"]

# Encode labels output
le=LabelEncoder()
y=le.fit_transform(labels)

# Encode input text
vectorizer = CountVectorizer()
X=vectorizer.fit_transform(emails)

# Train models
model=MultinomialNB()
model.fit(X,y)

# Test
test_email=["win free"]
test_vector=vectorizer.transform(test_email)
prediction=model.predict(test_vector)
print("Prediction Result using MultinomialNB: ",le.inverse_transform(prediction)[0])

```

```python
# GaussianNB

import numpy as np
from sklearn.naive_bayes import GaussianNB
from sklearn.preprocessing import LabelEncoder

X=np.array([[35],[45],[50],[65],[85]])

y_label=["Fail","Fail","Fail","Pass","Pass"]

# Encode labels(fail=0,pass=1)

le=LabelEncoder()
y=le.fit_transform(y_label)

model=GaussianNB()

model.fit(X,y)

# test model with new mark
new_mark=[[66]]

pred=model.predict(new_mark)
# Decode result for new mark
print("Result for New Mark [66] under GaussianNB :",le.inverse_transform(pred)[0])
```

```python
# Bernouli Naive Bayes predicts whether a person will:

# Buy a Laptop or Not
# based on:
# 1.Internet Usage
# 2.Office Work
# (Binary values → Yes/No → 1/0)


import numpy as np
from sklearn.naive_bayes import BernoulliNB
from sklearn.preprocessing import LabelEncoder

# input features (Intenet usage, Office work)
X=np.array([[0,0],  #not buy
            [0,1],  #not buy
            [1,1],  #buy
            [1,0]   #buy
            ])

y_label=np.array(["Not buy","Not buy","buy","buy"])

# encode y label
le=LabelEncoder()
y=le.fit_transform(y_label)

# model create and train model
model=BernoulliNB()
model.fit(X,y)

# predict for new input =[1,0]
new_data=[[1,0]]
pred=model.predict(new_data)

# Decode prediction
print("Prediction for new data using BernoulliNB: ",le.inverse_transform(pred)[0])
```
