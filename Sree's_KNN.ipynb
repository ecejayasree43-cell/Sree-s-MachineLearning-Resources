```python
# KNN -  K NEAREST NEIGBORS

# This algorith checks nearby data points and predicts based on majority voting

# This program teaches ML model to identify whether a fruit is an Apple or Orange

import numpy as np
from sklearn.neighbors import KNeighborsClassifier
```

```python
# Dataset :
# Weight (g), Texture - 0 for Smooth, 1 for Roughness
X=np.array([[145,0],[159,0],[200,1],[180,1] ])

# labels
y=np.array(["Apple","Orange","Apple","Orange"])

# Create KNN model (k=3)
model=KNeighborsClassifier(n_neighbors=3)

# Train model
model.fit(X,y)

# Predict New Fruit
new_fruit=[[158,0]] # where weight=158g and texture = 0 (smooth)

# make prediction
prediction=model.predict(new_fruit)
print("Prediction Result for New Fruit is ",prediction[0])

#  Hence predicted result (New fruit) is Orange based on KNN
```
